```python
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import (classification_report, f1_score,
                             accuracy_score, roc_auc_score,
                             precision_recall_curve, average_precision_score)
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# =========================
# 1. LOAD DATA
# =========================

df = pd.read_csv(
    r"C:\Users\Riya Bhardwaj\OneDrive - Flock Energy\SingaporeIM\output\New_IM_output.csv",
    parse_dates=["date_time"]
)

df = df.rename(columns={

    "AC_1_State": "ac1_state",
    "AC_2_State": "ac2_state",
    "AC_3_State": "ac3_state",

    "AC_1_Power_W": "ac1_power",
    "AC_2_Power_W": "ac2_power",
    "AC_3_Power_W": "ac3_power",

    "Cell_Tester_1_State": "cell_tester1_state",
    "Cell_Tester_1_Power_W": "cell_tester1_power",

    "Chamber_1_State": "chamber1_state",
    "Chamber_2_State": "chamber2_state",
    "Chamber_3_State": "chamber3_state",
    "Chamber_4_State": "chamber4_state",
    "Chamber_5_State": "chamber5_state",
    "Chamber_6_State": "chamber6_state",
    "Chamber_7_State": "chamber7_state",
    "Chamber_8_State": "chamber8_state",
    "Chamber_9_State": "chamber9_state",

    "Chamber_1_Power_W": "chamber1_power",
    "Chamber_2_Power_W": "chamber2_power",
    "Chamber_3_Power_W": "chamber3_power",
    "Chamber_4_Power_W": "chamber4_power",
    "Chamber_5_Power_W": "chamber5_power",
    "Chamber_6_Power_W": "chamber6_power",
    "Chamber_7_Power_W": "chamber7_power",
    "Chamber_8_Power_W": "chamber8_power",
    "Chamber_9_Power_W": "chamber9_power",
})

# Make sure house IDs are standardized
df["house_id"] = df["house_id"].replace({
    "House1_Jaipur": "House_1",
    "House2_Jaipur": "House_2",
    "House3_Jaipur": "House_4"
})

# Convert datetime column into proper datetime format
df["date_time"] = pd.to_datetime(
    df["date_time"],
    dayfirst=True,
    errors="coerce"
)

# Use only one house for modelling
df = df[df["house_id"] == "Singapore_2"].copy()

# Extract date for daily aggregation
df["date"] = df["date_time"].dt.date

print("Dataset loaded successfully.")
print("Shape:", df.shape)
print("\nHouse distribution:")
print(df["house_id"].value_counts())

print("\nDate range:")
print(df["date_time"].min(), "→", df["date_time"].max())

# =========================
# DATA VALIDATION
# =========================

required_columns = [
    "date_time",
    "house_id",
    "Temperature",
    "Humidity",

    "ac1_state",
    "ac1_power",

    "ac2_state",
    "ac2_power",

    "ac3_state",
    "ac3_power",

    "cell_tester1_state",
    "cell_tester1_power",

    "chamber1_state",
    "chamber1_power"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    print("\nMissing columns:")
    print(missing_columns)
else:
    print("\nAll required columns are present.")

print("\nMissing values:")
print(df[
    ["date_time", "house_id",
     "ac_state", "ac_power",
     "Temperature", "Humidity"]
].isna().sum())

# =========================
# 2. FEATURE ENGINEERING (added features related to weather)
# =========================
def extract_daily_features(group, appliance):
    power_col = f"{appliance}_power"
    state_col = f"{appliance}_state"
    on_data   = group[group[state_col] == 1]

    # -------------------------------------------------
    # DATA AVAILABILITY CHECK
    # -------------------------------------------------

    expected_intervals = 48

    actual_intervals = len(group)

    valid_power = group[power_col].notna().sum()
    valid_state = group[state_col].notna().sum()

    zero_power_count = (
        group[power_col].fillna(0).eq(0).sum()
    )

    # Percentage of zero-power readings
    zero_power_ratio = (
        zero_power_count / expected_intervals
    )

    # Basic structural validation
    if (
        actual_intervals < expected_intervals
        or valid_power < expected_intervals
        or valid_state < expected_intervals
    ):
        data_status = "MISSING_OR_PARTIAL"

    else:
        data_status = "VALID"

    base = {
        f"{appliance}_data_status":     data_status,
        f"{appliance}_raw_rows":        actual_intervals,
        f"{appliance}_valid_power_rows": valid_power,
        f"{appliance}_valid_state_rows": valid_state,
        f"{appliance}_total_energy":      0,
        f"{appliance}_on_duration":       0,
        f"{appliance}_energy_per_hour":   0,
        f"{appliance}_duty_cycle":        0,
        f"{appliance}_cycles":            0,
        f"{appliance}_std_power":         0,
        f"{appliance}_power_range":       0,
        f"{appliance}_cv_power":          0,
        f"{appliance}_short_cycles":      0,
        f"{appliance}_long_run_ratio":    0,
        f"{appliance}_power_gradient":    0,
        # Added weather features
        f"{appliance}_temperature_mean":  0,
        f"{appliance}_temperature_std":   0,
        f"{appliance}_temp_runtime":      0,
        f"{appliance}_humidity_mean":     0,
        f"{appliance}_humidity_std":      0,
        f"{appliance}_humidity_runtime":  0,
        f"{appliance}_runtime_per_degree":0,
        f"{appliance}_heat_index":        0,
        f"{appliance}_peak_average_ratio":0
    }

    # -------------------------------------------------
    # MISSING / PARTIAL DATA
    # -------------------------------------------------
    if data_status == "MISSING_OR_PARTIAL":
        return base

    # -------------------------------------------------
    # VALID DATA BUT APPLIANCE WAS OFF
    # -------------------------------------------------
    if len(on_data) == 0:
        return base

    energy = on_data[power_col].sum()*0.5  # Assuming 30-minute intervals, convert to kWh
    on_duration = len(on_data)*0.5      
    avg_power   = on_data[power_col].mean()
    peak_power = on_data[power_col].max()
    min_power = on_data[power_col].min()
    median_power = on_data[power_col].median()
    p90_power = on_data[power_col].quantile(0.90)
    p95_power = on_data[power_col].quantile(0.95)
    std_power   = on_data[power_col].std() if len(on_data) > 1 else 0

    temperature_mean = group["Temperature"].mean()
    humidity_mean = group["Humidity"].mean()
    peak_average_ratio = peak_power / avg_power
    power_gradient = (
        on_data[power_col]
        .diff()
        .abs()
        .mean()
    )

    if pd.isna(power_gradient):
        power_gradient = 0
    

    temperature_std = (
        group["Temperature"].std()
        if len(group) > 1 else 0
    )

    humidity_std = (
        group["Humidity"].std()
        if len(group) > 1 else 0
    )

    # Weather-normalized features

    runtime_per_degree = (
        on_duration / temperature_mean
        if temperature_mean > 0 else 0
    )

# Simple Heat Index
    heat_index = (
        temperature_mean + 0.1 * humidity_mean
    )
    temp_runtime = temperature_mean * on_duration
    humidity_runtime = humidity_mean * on_duration

    group_copy = group.copy()
    group_copy["state_shift"] = group_copy[state_col].diff().fillna(0)
    cycles = group_copy[group_copy["state_shift"] == 1].shape[0]

    # Run-length analysis
    run_lengths, current_run = [], 0
    for val in group[state_col]:
        if val == 1:
            current_run += 1
        else:
            if current_run > 0:
                run_lengths.append(current_run)
                current_run = 0
    if current_run > 0:
        run_lengths.append(current_run)

    short_cycles   = sum(1 for r in run_lengths if r < 10)
    long_run_ratio = (sum(r for r in run_lengths if r > 30) / on_duration
                      if on_duration > 0 else 0)

    base.update({
        f"{appliance}_total_energy":    energy,
        f"{appliance}_on_duration":     on_duration,
        f"{appliance}_energy_per_hour": energy / on_duration,
        f"{appliance}_duty_cycle":      group[state_col].mean(),
        f"{appliance}_cycles":          cycles,
        f"{appliance}_std_power":       std_power,
        f"{appliance}_power_range":     on_data[power_col].max() - on_data[power_col].min(),
        f"{appliance}_cv_power":        std_power / avg_power if avg_power > 0 else 0,
        f"{appliance}_short_cycles":    short_cycles,
        f"{appliance}_long_run_ratio":  long_run_ratio,
        f"{appliance}_temperature_mean": temperature_mean,    
        f"{appliance}_temperature_std": temperature_std,
        f"{appliance}_humidity_mean": humidity_mean,
        f"{appliance}_humidity_std": humidity_std,
        f"{appliance}_runtime_per_degree": runtime_per_degree,
        f"{appliance}_heat_index": heat_index,
        f"{appliance}_temp_runtime": temp_runtime,
        f"{appliance}_humidity_runtime": humidity_runtime,
        f"{appliance}_peak_average_ratio": peak_average_ratio,
        f"{appliance}_power_gradient": power_gradient,
        f"{appliance}_mean_power": avg_power,
        f"{appliance}_max_power": peak_power,
        f"{appliance}_median_power": median_power,
        f"{appliance}_p90_power": p90_power,
        f"{appliance}_p95_power": p95_power
    })
    return base

def build_daily_data(df, appliance):
    records = []
    for house in df["house_id"].unique():
        for date, day_group in df[df['house_id'] == house].groupby('date'):
            feat = extract_daily_features(day_group, appliance)
            feat.update({"house_id": house, "date": date})
            records.append(feat)
    return pd.DataFrame(records)

# =========================
# 3. STAR RATINGS
# =========================
meta = pd.read_csv(r"C:\Users\Riya Bhardwaj\OneDrive - Flock Energy\SingaporeIM\data\hosue_appliances_gt.csv")
def get_star_ratings(meta, appliance_type):

    subset = meta[
        meta["appliance_type"] == appliance_type
    ].dropna(subset=["star_rating"]).copy()

    if subset.empty:
        return pd.DataFrame(
            columns=["house_id", "star_rating"]
        )

    ratings = []

    for house_id, group in subset.groupby("house_id"):

        ratings.append({
            "house_id": house_id,
            "star_rating": np.average(
                group["star_rating"],
                weights=group["appliance_count"]
            )
        })

    return pd.DataFrame(ratings)
# =========================
# 4. RESIDUAL-BASED LABELING
# =========================
#
# Objective:
#   expected_energy = LinearRegression(...)
#   residual        = actual_energy - expected_energy
#
# IMPORTANT DATA HANDLING:
#
# VALID DAY
#   → residual is calculated normally
#
# VALID ZERO-ENERGY DAY
#   → expected_energy = 0
#   → residual = 0
#
# MISSING / PARTIAL / OUTAGE DAY
#   → expected_energy = NaN
#   → residual = NaN
#   → efficiency label = NaN
#
# Outage days are therefore excluded from:
#   - baseline regression training
#   - expected energy calculation
#   - residual threshold calculation
#   - efficiency labeling
#
# =========================


def assign_residual_labels(data, appliance, train_mask):

    dur_col = f"{appliance}_on_duration"
    energy_col = f"{appliance}_total_energy"
    status_col = f"{appliance}_data_status"

    data = data.copy()

    # -------------------------------------------------
    # 1. IDENTIFY INVALID / OUTAGE DAYS
    # -------------------------------------------------

    invalid_data_mask = data[status_col].isin([
        "MISSING_OR_PARTIAL",
        "OUTAGE_OR_NO_DATA"
    ])

    print("\n===== DATA STATUS CHECK =====")

    print(
        data[status_col]
        .value_counts(dropna=False)
        .to_string()
    )

    print(
        f"\nInvalid / outage days excluded: "
        f"{invalid_data_mask.sum()}"
    )

    # -------------------------------------------------
    # 2. ACTIVE + VALID DAYS
    # -------------------------------------------------
    #
    # These are the ONLY days allowed to train the
    # expected-energy regression model.
    #
    # Requirements:
    #   - appliance was active
    #   - actual energy > 0
    #   - data is valid
    #
    # -------------------------------------------------

    active_mask = (
        (data[dur_col] > 0) &
        (data[energy_col] > 0) &
        (~invalid_data_mask)
    )

    train_active = data[
        train_mask & active_mask
    ].copy()

    print(
        f"Valid active training days: "
        f"{len(train_active)}"
    )

    if len(train_active) < 5:

        print(
            "  WARNING: too few active train days "
            "for residual model"
        )

        return None

    # -------------------------------------------------
    # 3. BASELINE FEATURES
    # -------------------------------------------------

    baseline_features = [
        f"{appliance}_std_power",
        f"{appliance}_power_range",
        f"{appliance}_cv_power",
        f"{appliance}_cycles",
        f"{appliance}_heat_index"
    ]

    # -------------------------------------------------
    # 4. CLEAN BASELINE FEATURES
    # -------------------------------------------------

    data[baseline_features] = (
        data[baseline_features]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(0)
    )

    train_active[baseline_features] = (
        train_active[baseline_features]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(0)
    )

    # -------------------------------------------------
    # 5. TRAIN EXPECTED-ENERGY REGRESSION
    # -------------------------------------------------

    lr = LinearRegression()

    lr.fit(
        train_active[baseline_features],
        train_active[energy_col]
    )

    # -------------------------------------------------
    # 6. INITIALIZE EXPECTED ENERGY + RESIDUAL
    #
    # IMPORTANT:
    # Start everything as NaN.
    #
    # This means an invalid/outage day will remain
    # excluded unless we explicitly calculate it.
    # -------------------------------------------------

    data["expected_energy"] = np.nan
    data["energy_residual"] = np.nan

    # -------------------------------------------------
    # 7. PREDICT ONLY FOR VALID DAYS
    # -------------------------------------------------

    valid_prediction_mask = ~invalid_data_mask

    if valid_prediction_mask.any():

        data.loc[
            valid_prediction_mask,
            "expected_energy"
        ] = lr.predict(
            data.loc[
                valid_prediction_mask,
                baseline_features
            ]
        )

        # -------------------------------------------------
        # 8. CALCULATE RESIDUAL
        # -------------------------------------------------

        data.loc[
            valid_prediction_mask,
            "energy_residual"
        ] = (
            data.loc[
                valid_prediction_mask,
                energy_col
            ]
            -
            data.loc[
                valid_prediction_mask,
                "expected_energy"
            ]
        )

    # -------------------------------------------------
    # 9. VALID ZERO-ENERGY DAYS
    # -------------------------------------------------
    #
    # A valid zero-energy day is NOT an outage.
    #
    # Therefore:
    #
    # expected_energy = 0
    # residual        = 0
    #
    # -------------------------------------------------

    valid_zero_mask = (
        (data[energy_col] <= 0) &
        (~invalid_data_mask)
    )

    data.loc[
        valid_zero_mask,
        "expected_energy"
    ] = 0.0

    data.loc[
        valid_zero_mask,
        "energy_residual"
    ] = 0.0

    # -------------------------------------------------
    # 10. FORCE INVALID / OUTAGE DAYS TO NaN
    # -------------------------------------------------
    #
    # This is the critical outage protection.
    #
    # Even if the regression somehow generated a
    # prediction, an outage day must never have a
    # residual.
    #
    # -------------------------------------------------

    data.loc[
        invalid_data_mask,
        "expected_energy"
    ] = np.nan

    data.loc[
        invalid_data_mask,
        "energy_residual"
    ] = np.nan

    # -------------------------------------------------
    # 11. VALID RESIDUALS FOR THRESHOLD CALCULATION
    # -------------------------------------------------

    valid_residual_mask = (
        data["energy_residual"].notna()
    )

    train_residuals = data.loc[
        train_mask &
        active_mask &
        valid_residual_mask,
        "energy_residual"
    ]

    if len(train_residuals) < 5:

        print(
            f"  WARNING: insufficient valid residual days "
            f"for {appliance}"
        )

        return None

    # -------------------------------------------------
    # 12. BASE RESIDUAL THRESHOLD
    # -------------------------------------------------

    base_threshold = np.percentile(
        train_residuals,
        75
    )

    # -------------------------------------------------
    # 13. RESIDUAL DISTRIBUTION
    # -------------------------------------------------

    valid_residuals = data.loc[
        valid_residual_mask,
        "energy_residual"
    ]

    plt.figure(figsize=(7, 4))

    plt.hist(
        valid_residuals,
        bins=20,
        edgecolor="black"
    )

    plt.axvline(
        base_threshold,
        color="red",
        linestyle="--",
        linewidth=2,
        label="Threshold"
    )

    plt.text(
        base_threshold,
        plt.ylim()[1] * 0.9,
        f"Threshold = {base_threshold:.2f}",
        color="red"
    )

    plt.xlabel("Residual")
    plt.ylabel("Frequency")
    plt.title(
        f"{appliance.upper()} - Residual Distribution"
    )

    plt.legend()
    plt.show()

    # -------------------------------------------------
    # 14. STAR-ADJUSTED THRESHOLD
    # -------------------------------------------------

    max_stars = data["star_rating"].max()

    if pd.isna(max_stars) or max_stars <= 0:

        print(
            "  WARNING: invalid star rating. "
            "Using base threshold."
        )

        adj_thresh = pd.Series(
            base_threshold,
            index=data.index
        )

    else:

        star_factor = (
            data["star_rating"] / max_stars
        )

        adj_thresh = (
            base_threshold *
            (2 - star_factor)
        )

    # -------------------------------------------------
    # 15. REPORT REGRESSION PERFORMANCE
    # -------------------------------------------------

    train_r2 = lr.score(
        train_active[baseline_features],
        train_active[energy_col]
    )

    print(
        f"\n  Residual baseline R² : "
        f"{train_r2:.4f}"
    )

    print(
        f"  Base threshold (75pct): "
        f"{base_threshold:.2f}"
    )

    # -------------------------------------------------
    # 16. REPORT HOUSE-SPECIFIC THRESHOLD
    # -------------------------------------------------

    for house in data["house_id"].unique():

        house_rows = data[
            data["house_id"] == house
        ]

        if house_rows.empty:
            continue

        row = house_rows.iloc[0]

        if (
            pd.isna(max_stars)
            or max_stars <= 0
            or pd.isna(row["star_rating"])
        ):

            thr = base_threshold

        else:

            thr = (
                base_threshold *
                (
                    2 -
                    row["star_rating"] / max_stars
                )
            )

        print(
            f"    {house} "
            f"(★{row['star_rating']:.2f}) "
            f"→ threshold={thr:.2f}"
        )

    # -------------------------------------------------
    # 17. FINAL EFFICIENCY LABELS
    # -------------------------------------------------
    #
    # Valid days:
    #     residual > threshold → 1
    #     residual <= threshold → 0
    #
    # Invalid / outage:
    #     NaN
    #
    # -------------------------------------------------

    labels = pd.Series(
        np.nan,
        index=data.index,
        dtype=float
    )

    valid_label_mask = (
        data["energy_residual"].notna() &
        (~invalid_data_mask)
    )

    labels.loc[valid_label_mask] = (
        data.loc[
            valid_label_mask,
            "energy_residual"
        ]
        >
        adj_thresh.loc[valid_label_mask]
    ).astype(int)

    # -------------------------------------------------
    # 18. FINAL OUTAGE SANITY CHECK
    # -------------------------------------------------

    print("\n===== OUTAGE RESIDUAL SANITY CHECK =====")

    outage_rows = data[invalid_data_mask]

    print(
        f"Outage / invalid days: {len(outage_rows)}"
    )

    print(
        f"Outage days with residual: "
        f"{outage_rows['energy_residual'].notna().sum()}"
    )

    print(
        f"Outage days with expected energy: "
        f"{outage_rows['expected_energy'].notna().sum()}"
    )

    print(
        f"Outage days with efficiency label: "
        f"{labels.loc[invalid_data_mask].notna().sum()}"
    )

    # -------------------------------------------------
    # 19. FINAL SAFETY ASSERTION
    # -------------------------------------------------
    #
    # These should ALWAYS be zero.
    #
    # -------------------------------------------------

    assert (
        data.loc[
            invalid_data_mask,
            "energy_residual"
        ].notna().sum()
        == 0
    ), "ERROR: Outage days still have residuals!"

    assert (
        data.loc[
            invalid_data_mask,
            "expected_energy"
        ].notna().sum()
        == 0
    ), "ERROR: Outage days still have expected energy!"

    assert (
        labels.loc[
            invalid_data_mask
        ].notna().sum()
        == 0
    ), "ERROR: Outage days still have efficiency labels!"

    print(
        "✓ Outage handling passed sanity checks."
    )

    return labels, data["energy_residual"]

# =========================
# 5. MODEL FEATURES
# Only behavioral features — no energy, no avg_power, no usage_flag
# =========================

# Added temp_runtime along with heat_index to give model an eye on weather patterns 
def get_model_features(appliance):
    return [
        f"{appliance}_std_power",
        f"{appliance}_power_range",
        f"{appliance}_cv_power",
        f"{appliance}_temp_runtime",
        f"{appliance}_peak_average_ratio",
        f"{appliance}_power_gradient",
        f"{appliance}_heat_index",
    ]

# =========================
# 6. EVALUATION WITH PRECISION-RECALL
# Use PR-AUC as primary metric for imbalanced classes
# =========================
def evaluate(y_test, y_pred, y_prob, feature_cols, X_test, house, appliance):
    acc   = accuracy_score(y_test, y_pred)
    f1    = f1_score(y_test, y_pred, zero_division=0)
    n_pos = y_test.sum()

    try:
        roc = roc_auc_score(y_test, y_prob)
    except ValueError:
        roc = None

    try:
        pr_auc = average_precision_score(y_test, y_prob)
    except ValueError:
        pr_auc = None

    roc_str   = f"{roc:.4f}"   if roc   is not None else "N/A"
    prauc_str = f"{pr_auc:.4f}" if pr_auc is not None else "N/A"

    print(f"\n  Test positives : {n_pos} / {len(y_test)}")
    print(f"  Accuracy : {acc:.4f} | F1 : {f1:.4f} | "
          f"ROC-AUC : {roc_str} | PR-AUC : {prauc_str}")

    if n_pos < 3:
        print("   WARNING: fewer than 3 positives in test — "
              "F1/ROC unreliable. Treat PR-AUC as indicative only.")

    print(classification_report(y_test, y_pred, zero_division=0,
                                 target_names=["Efficient", "Inefficient"]))
    # ----------------------------
    # Confusion Matrix
    # ----------------------------
    cm = confusion_matrix(y_test, y_pred)

    print("\nConfusion Matrix:")
    print(cm)

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Efficient", "Inefficient"]
    )

    disp.plot(cmap="Blues")
    plt.title(f"{house} - {appliance.upper()} Confusion Matrix")
    plt.show()


    # Feature importance
    fi = pd.Series(dict(zip(feature_cols,
                            [0]*len(feature_cols)))).copy()
    return acc, f1, roc, pr_auc

# =========================
# 7. TEMPORAL SPLIT
# Train only on ACTIVE days — removes usage_flag leakage
# =========================
def run_temporal_split(data, appliance, train_ratio=0.7):
    dur_col      = f"{appliance}_on_duration"
    feature_cols = get_model_features(appliance)

    print(f"\n{'='*60}")
    print(f"  APPLIANCE : {appliance.upper()}")
    print(f"  Features  : {feature_cols}")
    print(f"{'='*60}")

    all_results = []

    for house in data["house_id"].unique():
        # Sort chronologically, keep ALL days for labeling context
        house_data = data[data["house_id"] == house].sort_values("date").copy()
        status_col = f"{appliance}_data_status"

        active_days = house_data[
            (house_data[dur_col] > 0) &
            (house_data[status_col] == "VALID")
        ]

        print(f"\n[{house}]")
        print(f"  Total days: {len(house_data)} | Active days: {len(active_days)}")

        if len(active_days) == 0:
            print("  SKIP — no activity")
            continue

        # Split ACTIVE days chronologically
        n         = len(active_days)
        split_idx = int(n * train_ratio)

        if split_idx < 5 or (n - split_idx) < 3:
            print(f"  SKIP — too few active days for split (active={n}, "
                  f"need train≥5, test≥3)")
            continue

        # Build train mask on full house_data index
        train_dates = set(active_days.iloc[:split_idx]["date"])
        test_dates  = set(active_days.iloc[split_idx:]["date"])

        train_mask = house_data["date"].isin(train_dates)
        test_mask  = house_data["date"].isin(test_dates)

        # Assign labels
        result = assign_residual_labels(house_data, appliance, train_mask)
        if result is None:
            continue

        labels, residuals = result
        house_data["efficiency_class"] = labels
        house_data["energy_residual"]  = residuals

        print("\n===== RESIDUAL SUMMARY =====")

        print(
            house_data["energy_residual"]
            .describe(percentiles=[0.50,0.75,0.90,0.95,0.99])
        )

        threshold = np.percentile(
            house_data.loc[train_mask &
                        (house_data[f"{appliance}_on_duration"] > 0),
                        "energy_residual"],
            75
        )

        print(f"\nThreshold Used = {threshold:.2f}")

        print(
            "\nDays Above Threshold:",
            (house_data["energy_residual"] > threshold).sum()
        )

        print(
            "Negative Residual Days:",
            (house_data["energy_residual"] < 0).sum()
        )

        print("\n===== TOP 10 POSITIVE RESIDUAL DAYS =====")

        print(
            house_data.sort_values(
                "energy_residual",
                ascending=False
            )[
                [
                    "date",
                    "energy_residual",
                    f"{appliance}_total_energy",
                    f"{appliance}_on_duration"
                ]
            ].head(10)
        )

        print("\n===== TOP 10 NEGATIVE RESIDUAL DAYS =====")

        print(
            house_data.sort_values(
                "energy_residual"
            )[
                [
                    "date",
                    "energy_residual",
                    f"{appliance}_total_energy",
                    f"{appliance}_on_duration"
                ]
            ].head(10)
        )

        train_data = house_data[train_mask]
        test_data  = house_data[test_mask]
        y_train    = train_data["efficiency_class"]
        y_test     = test_data["efficiency_class"]

# ----------------------------
# Label Distribution
# ----------------------------
        label_counts = house_data["efficiency_class"].value_counts().sort_index()

        print("\nLabel Distribution:")
        print(label_counts)

        plt.figure(figsize=(5,4))
        bars = plt.bar(
            ["Efficient", "Inefficient"],
            label_counts.values,
            color=["green", "red"]
        )

        for bar in bars:
            plt.text(
                bar.get_x()+bar.get_width()/2,
                bar.get_height()+0.3,
                int(bar.get_height()),
                ha="center"
            )

        plt.title(f"{house} - Label Distribution")
        plt.ylabel("No. of Days")
        plt.show()

        print(f"  Train active: {len(train_data)} days "
              f"({train_data['date'].min()} → {train_data['date'].max()}) "
              f"| classes: {y_train.value_counts().to_dict()}")
        print(f"  Test  active: {len(test_data)} days "
              f"({test_data['date'].min()} → {test_data['date'].max()}) "
              f"| classes: {y_test.value_counts().to_dict()}")

        if y_train.nunique() < 2:
            print("  SKIP — single class in train")
            continue
        if y_test.sum() == 0:
            print("  SKIP — no positive samples in test set")
            continue

        X_train = train_data[feature_cols]
        X_test  = test_data[feature_cols]

        scale_pos_weight = (
            (y_train == 0).sum()
            /
            (y_train == 1).sum()
        )

        model = XGBClassifier(
            n_estimators=300, 
            learning_rate=0.1,
            max_depth=4,
            min_child_weight=4,
            subsample=1.0,
            colsample_bytree=1.0,
            gamma=1,
            reg_alpha=0.5,
            reg_lambda=2,
            scale_pos_weight=scale_pos_weight,
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42
        )
        model.fit(X_train, y_train)

        from sklearn.metrics import accuracy_score, f1_score

        # Training predictions
        y_train_pred = model.predict(X_train)

        print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
        print("Train F1:", f1_score(y_train, y_train_pred))

        # Test predictions
        y_test_pred = model.predict(X_test)

        print("Test Accuracy:", accuracy_score(y_test, y_test_pred))
        print("Test F1:", f1_score(y_test, y_test_pred))

        y_pred = model.predict(X_test)
        y_prob = (model.predict_proba(X_test)[:, 1]
                  if len(model.classes_) > 1 else np.zeros(len(X_test)))
        
        test_results = test_data.copy()

        test_results["Actual"] = y_test.values
        test_results["Predicted"] = y_pred
        test_results["Probability"] = y_prob

        false_positive = test_results[
            (test_results["Actual"] == 0) &
            (test_results["Predicted"] == 1)
        ]

        false_negative = test_results[
            (test_results["Actual"] == 1) &
            (test_results["Predicted"] == 0)
        ]

        print("\nFalse Positives")
        print(false_positive)

        print("\nFalse Negatives")
        print(false_negative)
        print("\n===== FP Feature Mean =====")
        print(false_positive[feature_cols].mean())

        print("\n===== FN Feature Mean =====")
        print(false_negative[feature_cols].mean())

        print("\n===== TP Feature Mean =====")
        print(test_results[
            (test_results["Actual"]==1)&
            (test_results["Predicted"]==1)
        ][feature_cols].mean())

        print("\n===== TN Feature Mean =====")
        print(test_results[
            (test_results["Actual"]==0)&
            (test_results["Predicted"]==0)
        ][feature_cols].mean())

        acc, f1, roc, pr_auc = evaluate(
            y_test, y_pred, y_prob, feature_cols, X_test, house, appliance
        )
 
        from sklearn.metrics import roc_curve

        fpr, tpr, _ = roc_curve(y_test, y_prob)

        plt.figure(figsize=(6,5))
        plt.plot(fpr, tpr, label=f"AUC={roc:.3f}")
        plt.plot([0,1],[0,1],"--")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title("ROC Curve")
        plt.legend()
        plt.show()

        # Feature importance
        fi = pd.Series(model.feature_importances_,
                       index=feature_cols).sort_values(ascending=False)

        plt.figure(figsize=(8,5))
        fi.plot(kind="bar")
        plt.title(f"{house} - Feature Importance")
        plt.ylabel("Importance")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

        print(f"  Feature Importance:\n{fi.round(4).to_string()}")
        
        threshold = np.percentile(
            house_data.loc[
                train_mask &
                (house_data[f"{appliance}_on_duration"] > 0),
                "energy_residual"
            ],
            75
        )

        high_residual_days = house_data[
            house_data["energy_residual"] > threshold
        ]

        normal_days = house_data[
            house_data["energy_residual"] <= threshold
        ]

        print("\n===== HIGH RESIDUAL VS NORMAL DAYS =====")

        comparison = pd.DataFrame({
            "High Residual": high_residual_days[feature_cols].mean(),
            "Normal": normal_days[feature_cols].mean()
        })

        print(comparison)

        plt.figure(figsize=(8,5))

        plt.hist(
            house_data["energy_residual"],
            bins=30,
            edgecolor="black"
        )

        plt.axvline(
            threshold,
            color="red",
            linestyle="--",
            linewidth=2,
            label=f"Threshold = {threshold:.2f}"
        )

        plt.title(
            f"{house.upper()} - {appliance.upper()} Residual Distribution"
        )

        plt.xlabel("Residual")
        plt.ylabel("Count")
        plt.legend()

        plt.show()

        # Correlation check — should all be < 0.7 now
        eph = f"{appliance}_energy_per_hour"
        corr = house_data[feature_cols + [eph]].corr()[eph].drop(eph)

        high_corr = corr[corr.abs() > 0.7]
        if len(high_corr) > 0:
            print(f"\n   High correlation with energy_per_hour (>0.7):")
            print(high_corr.round(4).to_string())
        else:
            print(f"\n   All features correlation with energy_per_hour < 0.7")
        all_results.append({
            "appliance":           appliance,
            "house":               house,
            "active_train_days":   len(train_data),
            "active_test_days":    len(test_data),
            "test_positives":      int(y_test.sum()),
            "train_pos_rate":      round(y_train.mean(), 3),
            "test_pos_rate":       round(y_test.mean(), 3),
            "accuracy":            round(acc, 4),
            "f1_score":            round(f1, 4),
            "roc_auc":             round(roc, 4) if roc else None,
            "pr_auc":              round(pr_auc, 4) if pr_auc else None,
        })

    return all_results

# =========================
# 8. RUN ALL APPLIANCE MODELS
# =========================

appliance_map = {

    "ac1": ("AC", ["Singapore_2"]),
    "ac2": ("AC", ["Singapore_2"]),
    "ac3": ("AC", ["Singapore_2"]),

    "cell_tester1": ("Cell Tester", ["Singapore_2"]),

    "chamber1": ("Chamber", ["Singapore_2"]),
    "chamber2": ("Chamber", ["Singapore_2"]),
    "chamber3": ("Chamber", ["Singapore_2"]),
    "chamber4": ("Chamber", ["Singapore_2"]),
    "chamber5": ("Chamber", ["Singapore_2"]),
    "chamber6": ("Chamber", ["Singapore_2"]),
    "chamber7": ("Chamber", ["Singapore_2"]),
    "chamber8": ("Chamber", ["Singapore_2"]),
    "chamber9": ("Chamber", ["Singapore_2"])
}

# -------------------------------------------------
# TEMPORARY STAR RATINGS
# -------------------------------------------------
# IMPORTANT:
# These are ONLY placeholders for testing.
# Actual star ratings for Chamber and Cell Tester
# are currently unknown.

temporary_star_ratings = {

    ("ac1", "Singapore_2"): 3,
    ("ac2", "Singapore_2"): 3,
    ("ac3", "Singapore_2"): 3,

    ("cell_tester1", "Singapore_2"): 3,

    ("chamber1", "Singapore_2"): 3,
    ("chamber2", "Singapore_2"): 3,
    ("chamber3", "Singapore_2"): 3,
    ("chamber4", "Singapore_2"): 3,
    ("chamber5", "Singapore_2"): 3,
    ("chamber6", "Singapore_2"): 3,
    ("chamber7", "Singapore_2"): 3,
    ("chamber8", "Singapore_2"): 3,
    ("chamber9", "Singapore_2"): 3,
}


all_results = []


for appliance, (meta_type, active_houses) in appliance_map.items():

    print("\n" + "=" * 70)
    print(f"RUNNING {appliance.upper()} MODEL")
    print("=" * 70)

    # -------------------------------------------------
    # GET STAR RATINGS FROM METADATA
    # -------------------------------------------------

    ratings = get_star_ratings(
        meta,
        meta_type
    )

    # -------------------------------------------------
    # SELECT RELEVANT HOUSES
    # -------------------------------------------------

    df_sub = df[
        df["house_id"].isin(active_houses)
    ].copy()

    if df_sub.empty:

        print(
            f"\nSKIPPING {appliance}: "
            f"No data found for houses {active_houses}"
        )

        continue

    # -------------------------------------------------
    # BUILD DAILY FEATURES
    # -------------------------------------------------

    daily = build_daily_data(
        df_sub,
        appliance
    )

    # -------------------------------------------------
    # MERGE STAR RATINGS FROM METADATA
    # -------------------------------------------------

    daily = daily.merge(
        ratings,
        on="house_id",
        how="left"
    )

    # -------------------------------------------------
    # APPLY TEMPORARY STAR RATINGS
    # -------------------------------------------------

    for (app, house), rating in temporary_star_ratings.items():

        if appliance == app:

            daily.loc[
                daily["house_id"] == house,
                "star_rating"
            ] = rating

    # -------------------------------------------------
    # FILL REMAINING MISSING RATINGS
    # -------------------------------------------------

    daily["star_rating"] = (
        daily
        .groupby("house_id")["star_rating"]
        .transform(
            lambda x: x.fillna(x.median())
        )
    )

    # -------------------------------------------------
    # CHECK FOR MISSING RATINGS
    # -------------------------------------------------

    if daily["star_rating"].isna().any():

        missing_houses = (
            daily.loc[
                daily["star_rating"].isna(),
                "house_id"
            ]
            .unique()
        )

        print(
            f"\nWARNING: Missing star rating for "
            f"{appliance} in houses: "
            f"{list(missing_houses)}"
        )

        print(
            "Please provide a temporary or actual "
            "star rating before running this model."
        )

        continue

    # -------------------------------------------------
    # DISPLAY STAR RATINGS
    # -------------------------------------------------

    print("\nStar ratings being used:")

    print(
        daily[
            ["house_id", "star_rating"]
        ]
        .drop_duplicates()
        .sort_values("house_id")
        .to_string(index=False)
    )

    # -------------------------------------------------
    # RUN TEMPORAL SPLIT
    # -------------------------------------------------

    results = run_temporal_split(
        daily,
        appliance
    )

    all_results.extend(results)

# =========================
# 9. FINAL SUMMARY
# =========================
print("\n" + "="*60)
print("FINAL SUMMARY")
print("="*60)
if all_results:
    results_df = pd.DataFrame(all_results)
    print(results_df.to_string(index=False))
    print(f"\n{'─'*40}")
    print("Interpretation guide:")
    print("  ROC-AUC > 0.75 = model ranks inefficient days well")
    print("  PR-AUC  > 0.50 = model is useful for imbalanced detection")
    print("  F1 < 0.5 with test_positives < 3 = data too sparse to judge")
else:
    print("No valid folds.")
```
