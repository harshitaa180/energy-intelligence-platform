```python
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import (classification_report, f1_score,
                             accuracy_score, roc_auc_score,
                             precision_recall_curve, average_precision_score)
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# =========================
# 1. LOAD DATA
# =========================
df = pd.read_csv(r"C:/Users/KIIT/Desktop/Inefficiency model/merged_df.csv",parse_dates=["DATE_TIME"])

# Rename to match the rest of your pipeline
df.rename(columns={"DATE_TIME": "date_time"}, inplace=True)

# Rename house IDs
df["house_id"] = df["house_id"].replace({
    "House1_Jaipur": "House_1",
    "House2_Jaipur": "House_2",
    "House3_Jaipur": "House_4"
})                
df = df[df['house_id'].isin(['House_1', 'House_2', 'House_4'])].copy()
df.rename(columns={'timestamp': 'date_time'}, inplace=True)
df['date'] = df['date_time'].dt.date

# =========================
# 2. FEATURE ENGINEERING (added features related to weather)
# =========================
def extract_daily_features(group, appliance):
    power_col = f"{appliance}_power"
    state_col = f"{appliance}_state"
    on_data   = group[group[state_col] == 1]

    base = {
        f"{appliance}_total_energy":      0,
        f"{appliance}_on_duration":       0,
        f"{appliance}_energy_per_hour":   0,
        f"{appliance}_duty_cycle":        0,
        f"{appliance}_cycles":            0,
        f"{appliance}_std_power":         0,
        f"{appliance}_power_range":       0,
        f"{appliance}_cv_power":          0,
        f"{appliance}_short_cycles":      0,
        f"{appliance}_long_run_ratio":    0,
        f"{appliance}_power_gradient":    0,
        # Added weather features
        f"{appliance}_temperature_mean":  0,
        f"{appliance}_temperature_std":   0,
        f"{appliance}_temp_runtime":      0,
        f"{appliance}_humidity_mean":     0,
        f"{appliance}_humidity_std":      0,
        f"{appliance}_humidity_runtime":  0,
        f"{appliance}_runtime_per_degree":0,
        f"{appliance}_heat_index":        0,
        f"{appliance}_peak_average_ratio":0
    }

    if len(on_data) == 0:
        return base

    energy      = on_data[power_col].sum()
    on_duration = len(on_data)
    avg_power   = on_data[power_col].mean()
    peak_power = on_data[power_col].max()
    min_power = on_data[power_col].min()
    std_power   = on_data[power_col].std() if len(on_data) > 1 else 0

    temperature_mean = group["Temperature"].mean()
    humidity_mean = group["Humidity"].mean()
    peak_average_ratio = peak_power / avg_power
    power_gradient = (
        on_data[power_col]
        .diff()
        .abs()
        .mean()
    )

    if pd.isna(power_gradient):
        power_gradient = 0
    

    temperature_std = (
        group["Temperature"].std()
        if len(group) > 1 else 0
    )

    humidity_std = (
        group["Humidity"].std()
        if len(group) > 1 else 0
    )

    # Weather-normalized features

    runtime_per_degree = (
        on_duration / temperature_mean
        if temperature_mean > 0 else 0
    )

# Simple Heat Index
    heat_index = (
        temperature_mean + 0.1 * humidity_mean
    )
    temp_runtime = temperature_mean * on_duration
    humidity_runtime = humidity_mean * on_duration

    group_copy = group.copy()
    group_copy["state_shift"] = group_copy[state_col].diff().fillna(0)
    cycles = group_copy[group_copy["state_shift"] == 1].shape[0]

    # Run-length analysis
    run_lengths, current_run = [], 0
    for val in group[state_col]:
        if val == 1:
            current_run += 1
        else:
            if current_run > 0:
                run_lengths.append(current_run)
                current_run = 0
    if current_run > 0:
        run_lengths.append(current_run)

    short_cycles   = sum(1 for r in run_lengths if r < 10)
    long_run_ratio = (sum(r for r in run_lengths if r > 30) / on_duration
                      if on_duration > 0 else 0)

    base.update({
        f"{appliance}_total_energy":    energy,
        f"{appliance}_on_duration":     on_duration,
        f"{appliance}_energy_per_hour": energy / on_duration,
        f"{appliance}_duty_cycle":      group[state_col].mean(),
        f"{appliance}_cycles":          cycles,
        f"{appliance}_std_power":       std_power,
        f"{appliance}_power_range":     on_data[power_col].max() - on_data[power_col].min(),
        f"{appliance}_cv_power":        std_power / avg_power if avg_power > 0 else 0,
        f"{appliance}_short_cycles":    short_cycles,
        f"{appliance}_long_run_ratio":  long_run_ratio,
        f"{appliance}_temperature_mean": temperature_mean,    
        f"{appliance}_temperature_std": temperature_std,
        f"{appliance}_humidity_mean": humidity_mean,
        f"{appliance}_humidity_std": humidity_std,
        f"{appliance}_runtime_per_degree": runtime_per_degree,
        f"{appliance}_heat_index": heat_index,
        f"{appliance}_temp_runtime": temp_runtime,
        f"{appliance}_humidity_runtime": humidity_runtime,
        f"{appliance}_peak_average_ratio": peak_average_ratio,
        f"{appliance}_power_gradient": power_gradient,
    })
    return base

def build_daily_data(df, appliance):
    records = []
    for house in df["house_id"].unique():
        for date, day_group in df[df['house_id'] == house].groupby('date'):
            feat = extract_daily_features(day_group, appliance)
            feat.update({"house_id": house, "date": date})
            records.append(feat)
    return pd.DataFrame(records)

# =========================
# 3. STAR RATINGS
# =========================
meta = pd.read_csv(r"C:/Users/KIIT/Desktop/Inefficiency model/hosue_appliances_gt (1).csv")

def get_star_ratings(meta, appliance_type):
    subset = meta[meta['appliance_type'] == appliance_type].dropna(subset=['star_rating'])
    ratings = (
        subset.groupby('house_id')
        .apply(lambda g: np.average(g['star_rating'], weights=g['appliance_count']),
               include_groups=False)
        .reset_index()
    )
    ratings.columns = ['house_id', 'star_rating']
    return ratings

# =========================
# 4. RESIDUAL-BASED LABELING
# expected_energy = LinearRegression(on_duration, duty_cycle, cycles, heat_index)
# residual = actual - expected
# label = 1 if residual > star-adjusted 75th percentile (train only)
# =========================
def assign_residual_labels(data, appliance, train_mask):
    dur_col    = f"{appliance}_on_duration"
    energy_col = f"{appliance}_total_energy"

    # Work on active days only — removes usage_flag trivial correlation
    active_mask  = data[dur_col] > 0
    train_active = data[train_mask & active_mask].copy()

    if len(train_active) < 5:
        print("  WARNING: too few active train days for residual model")
        return None
    
    # Added heat_index in baseline for weather integration 
    baseline_features = [
        f"{appliance}_on_duration",
        f"{appliance}_duty_cycle",
        f"{appliance}_cycles",
        f"{appliance}_heat_index",
    ]

    lr = LinearRegression()
    lr.fit(train_active[baseline_features], train_active[energy_col])

    data = data.copy()
    data["expected_energy"] = lr.predict(data[baseline_features])
    data["energy_residual"] = data[energy_col] - data["expected_energy"]

    # Threshold from train active days only
    train_residuals = data[train_mask & active_mask]["energy_residual"]
    base_threshold  = np.percentile(train_residuals, 75)

    plt.figure(figsize=(7,4))

    plt.hist(
        data["energy_residual"],
        bins=20,
        edgecolor="black",
    )

    plt.axvline(
        base_threshold,
        color="red",
        linestyle="--",
        linewidth=2,
        label="Threshold"
    )

    plt.text(
        base_threshold,
        plt.ylim()[1]*0.9,
        f"Threshold = {base_threshold:.2f}",
        color="red"
    )

    plt.xlabel("Residual")
    plt.ylabel("Frequency")
    plt.title("Residual Distribution")
    plt.legend()
    plt.show()


    max_stars   = data["star_rating"].max()
    star_factor = data["star_rating"] / max_stars
    adj_thresh  = base_threshold * (2 - star_factor)

    print(f"  Residual baseline R²  : "
          f"{lr.score(train_active[baseline_features], train_active[energy_col]):.4f}")
    print(f"  Base threshold (75pct): {base_threshold:.2f}")
    for house in data["house_id"].unique():
        row = data[data["house_id"] == house].iloc[0]
        thr = base_threshold * (2 - row["star_rating"] / max_stars)
        print(f"    {house} (★{row['star_rating']:.2f}) → threshold={thr:.2f}")

    labels = (data["energy_residual"] > adj_thresh).astype(int)
    return labels, data["energy_residual"]

# =========================
# 5. MODEL FEATURES
# Only behavioral features — no energy, no avg_power, no usage_flag
# =========================

# Added temp_runtime along with heat_index to give model an eye on weather patterns 
def get_model_features(appliance):
    return [
        f"{appliance}_duty_cycle",
        f"{appliance}_std_power",
        f"{appliance}_power_range",
        f"{appliance}_cv_power",
        f"{appliance}_short_cycles",
        f"{appliance}_temp_runtime",
        f"{appliance}_heat_index",
        f"{appliance}_peak_average_ratio",
        f"{appliance}_power_gradient",
    ]

# =========================
# 6. EVALUATION WITH PRECISION-RECALL
# Use PR-AUC as primary metric for imbalanced classes
# =========================
def evaluate(y_test, y_pred, y_prob, feature_cols, X_test, house, appliance):
    acc   = accuracy_score(y_test, y_pred)
    f1    = f1_score(y_test, y_pred, zero_division=0)
    n_pos = y_test.sum()

    try:
        roc = roc_auc_score(y_test, y_prob)
    except ValueError:
        roc = None

    try:
        pr_auc = average_precision_score(y_test, y_prob)
    except ValueError:
        pr_auc = None

    roc_str   = f"{roc:.4f}"   if roc   is not None else "N/A"
    prauc_str = f"{pr_auc:.4f}" if pr_auc is not None else "N/A"

    print(f"\n  Test positives : {n_pos} / {len(y_test)}")
    print(f"  Accuracy : {acc:.4f} | F1 : {f1:.4f} | "
          f"ROC-AUC : {roc_str} | PR-AUC : {prauc_str}")

    if n_pos < 3:
        print("   WARNING: fewer than 3 positives in test — "
              "F1/ROC unreliable. Treat PR-AUC as indicative only.")

    print(classification_report(y_test, y_pred, zero_division=0,
                                 target_names=["Efficient", "Inefficient"]))
    # ----------------------------
    # Confusion Matrix
    # ----------------------------
    cm = confusion_matrix(y_test, y_pred)

    print("\nConfusion Matrix:")
    print(cm)

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Efficient", "Inefficient"]
    )

    disp.plot(cmap="Blues")
    plt.title(f"{house} - {appliance.upper()} Confusion Matrix")
    plt.show()


    # Feature importance
    fi = pd.Series(dict(zip(feature_cols,
                            [0]*len(feature_cols)))).copy()
    return acc, f1, roc, pr_auc

# =========================
# 7. TEMPORAL SPLIT
# Train only on ACTIVE days — removes usage_flag leakage
# =========================
def run_temporal_split(data, appliance, train_ratio=0.7):
    dur_col      = f"{appliance}_on_duration"
    feature_cols = get_model_features(appliance)

    print(f"\n{'='*60}")
    print(f"  APPLIANCE : {appliance.upper()}")
    print(f"  Features  : {feature_cols}")
    print(f"{'='*60}")

    all_results = []

    for house in data["house_id"].unique():
        # Sort chronologically, keep ALL days for labeling context
        house_data = data[data["house_id"] == house].sort_values("date").copy()
        active_days = house_data[house_data[dur_col] > 0]

        print(f"\n[{house}]")
        print(f"  Total days: {len(house_data)} | Active days: {len(active_days)}")

        if len(active_days) == 0:
            print("  SKIP — no activity")
            continue

        # Split ACTIVE days chronologically
        n         = len(active_days)
        split_idx = int(n * train_ratio)

        if split_idx < 5 or (n - split_idx) < 3:
            print(f"  SKIP — too few active days for split (active={n}, "
                  f"need train≥5, test≥3)")
            continue

        # Build train mask on full house_data index
        train_dates = set(active_days.iloc[:split_idx]["date"])
        test_dates  = set(active_days.iloc[split_idx:]["date"])

        train_mask = house_data["date"].isin(train_dates)
        test_mask  = house_data["date"].isin(test_dates)

        # Assign labels
        result = assign_residual_labels(house_data, appliance, train_mask)
        if result is None:
            continue

        labels, residuals = result
        house_data["efficiency_class"] = labels
        house_data["energy_residual"]  = residuals

        train_data = house_data[train_mask]
        test_data  = house_data[test_mask]
        y_train    = train_data["efficiency_class"]
        y_test     = test_data["efficiency_class"]

# ----------------------------
# Label Distribution
# ----------------------------
        label_counts = house_data["efficiency_class"].value_counts().sort_index()

        print("\nLabel Distribution:")
        print(label_counts)

        plt.figure(figsize=(5,4))
        bars = plt.bar(
            ["Efficient", "Inefficient"],
            label_counts.values,
            color=["green", "red"]
        )

        for bar in bars:
            plt.text(
                bar.get_x()+bar.get_width()/2,
                bar.get_height()+0.3,
                int(bar.get_height()),
                ha="center"
            )

        plt.title(f"{house} - Label Distribution")
        plt.ylabel("No. of Days")
        plt.show()

        print(f"  Train active: {len(train_data)} days "
              f"({train_data['date'].min()} → {train_data['date'].max()}) "
              f"| classes: {y_train.value_counts().to_dict()}")
        print(f"  Test  active: {len(test_data)} days "
              f"({test_data['date'].min()} → {test_data['date'].max()}) "
              f"| classes: {y_test.value_counts().to_dict()}")

        if y_train.nunique() < 2:
            print("  SKIP — single class in train")
            continue
        if y_test.sum() == 0:
            print("  SKIP — no positive samples in test set")
            continue

        X_train = train_data[feature_cols]
        X_test  = test_data[feature_cols]

        model = XGBClassifier(
            n_estimators=300, 
            learning_rate=0.1,
            max_depth=4,
            min_child_weight=4,
            subsample=1.0,
            colsample_bytree=1.0,
            gamma=1,
            reg_alpha=0.5,
            reg_lambda=2,
            scale_pos_weight=2.8,
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42
        )
        model.fit(X_train, y_train)

        from sklearn.metrics import accuracy_score, f1_score

        # Training predictions
        y_train_pred = model.predict(X_train)

        print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
        print("Train F1:", f1_score(y_train, y_train_pred))

        # Test predictions
        y_test_pred = model.predict(X_test)

        print("Test Accuracy:", accuracy_score(y_test, y_test_pred))
        print("Test F1:", f1_score(y_test, y_test_pred))

        y_pred = model.predict(X_test)
        y_prob = (model.predict_proba(X_test)[:, 1]
                  if len(model.classes_) > 1 else np.zeros(len(X_test)))
        
        test_results = test_data.copy()

        test_results["Actual"] = y_test.values
        test_results["Predicted"] = y_pred
        test_results["Probability"] = y_prob

        false_positive = test_results[
            (test_results["Actual"] == 0) &
            (test_results["Predicted"] == 1)
        ]

        false_negative = test_results[
            (test_results["Actual"] == 1) &
            (test_results["Predicted"] == 0)
        ]

        print("\nFalse Positives")
        print(false_positive)

        print("\nFalse Negatives")
        print(false_negative)
        print("\n===== FP Feature Mean =====")
        print(false_positive[feature_cols].mean())

        print("\n===== FN Feature Mean =====")
        print(false_negative[feature_cols].mean())

        print("\n===== TP Feature Mean =====")
        print(test_results[
            (test_results["Actual"]==1)&
            (test_results["Predicted"]==1)
        ][feature_cols].mean())

        print("\n===== TN Feature Mean =====")
        print(test_results[
            (test_results["Actual"]==0)&
            (test_results["Predicted"]==0)
        ][feature_cols].mean())

        acc, f1, roc, pr_auc = evaluate(
            y_test, y_pred, y_prob, feature_cols, X_test, house, appliance
        )
 
        from sklearn.metrics import roc_curve

        fpr, tpr, _ = roc_curve(y_test, y_prob)

        plt.figure(figsize=(6,5))
        plt.plot(fpr, tpr, label=f"AUC={roc:.3f}")
        plt.plot([0,1],[0,1],"--")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title("ROC Curve")
        plt.legend()
        plt.show()

        # Feature importance
        fi = pd.Series(model.feature_importances_,
                       index=feature_cols).sort_values(ascending=False)

        plt.figure(figsize=(8,5))
        fi.plot(kind="bar")
        plt.title(f"{house} - Feature Importance")
        plt.ylabel("Importance")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

        print(f"  Feature Importance:\n{fi.round(4).to_string()}")

        # Correlation check — should all be < 0.7 now
        eph = f"{appliance}_energy_per_hour"
        corr = house_data[feature_cols + [eph]].corr()[eph].drop(eph)

        high_corr = corr[corr.abs() > 0.7]
        if len(high_corr) > 0:
            print(f"\n   High correlation with energy_per_hour (>0.7):")
            print(high_corr.round(4).to_string())
        else:
            print(f"\n   All features correlation with energy_per_hour < 0.7")
        all_results.append({
            "appliance":           appliance,
            "house":               house,
            "active_train_days":   len(train_data),
            "active_test_days":    len(test_data),
            "test_positives":      int(y_test.sum()),
            "train_pos_rate":      round(y_train.mean(), 3),
            "test_pos_rate":       round(y_test.mean(), 3),
            "accuracy":            round(acc, 4),
            "f1_score":            round(f1, 4),
            "roc_auc":             round(roc, 4) if roc else None,
            "pr_auc":              round(pr_auc, 4) if pr_auc else None,
        })

    return all_results

# =========================
# 8. RUN BOTH APPLIANCES
# =========================
appliance_map = {
    "ac":     ("AC",     ['House_4']),
    "geyser": ("Geyser", ['House_1', 'House_2'])
}

all_results = []
for appliance, (meta_type, active_houses) in appliance_map.items():
    ratings = get_star_ratings(meta, meta_type)
    df_sub  = df[df['house_id'].isin(active_houses)]
    daily   = build_daily_data(df_sub, appliance)
    daily   = daily.merge(ratings, on="house_id", how="left")
    daily["star_rating"] = daily.groupby("house_id")["star_rating"].transform(
        lambda x: x.fillna(x.median())
    )
    results = run_temporal_split(daily, appliance)
    all_results.extend(results)

# =========================
# 9. FINAL SUMMARY
# =========================
print("\n" + "="*60)
print("FINAL SUMMARY")
print("="*60)
if all_results:
    results_df = pd.DataFrame(all_results)
    print(results_df.to_string(index=False))
    print(f"\n{'─'*40}")
    print("Interpretation guide:")
    print("  ROC-AUC > 0.75 = model ranks inefficient days well")
    print("  PR-AUC  > 0.50 = model is useful for imbalanced detection")
    print("  F1 < 0.5 with test_positives < 3 = data too sparse to judge")
else:
    print("No valid folds.")
```
